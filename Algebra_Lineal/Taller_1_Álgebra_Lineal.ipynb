## Try-Except
Es una estructura para manejar errores. Funciona así:


```python
try:
    # Código que podría generar un error
    resultado = operacion_riesgosa()
except TipoDeError:
    # Qué hacer si ocurre ese error
    print("Ocurrió un error")
```
1. Python intenta ejecutar el código dentro del bloque `try`.
2. Si no hay errores, salta el `except`.
3. Si ocurre un error del tipo especificado, ejecuta el bloque `except`.

````python
print("A + D:")
try:
    print(A + D)
except ValueError as e:
    print(f"No definida: {e}")
````

Para clasificar una matriz $A \in \mathbb{R}^{n \times n}$ en términos de su estabilidad numérica y solubilidad, se consideran lo siguiente:

- **Matriz Singular**

    Una matriz se denomina singular si su determinante es nulo ($\det(A) = 0$). Esto implica que:
    - No existe su matriz inversa ($A^{-1}$)
    - El sistema lineal $A\mathbf{x} = \mathbf{b}$ puede tener infinitas soluciones o ninguna solución.


- **Matriz mal condicionada**

    Una matriz es mal condicionada cuando:
    - Su determinante es cercano a cero ($|\det(A)| \approx 0$).
    - Su **número de condición** $\kappa(A)$ es muy grande.

Estas matrices son numéricamente inestables: pequeñas perturbaciones en los datos pueden generar grandes cambios en la solución.

- **Matriz bien condicionada**

    Una matriz es bien condicionada cuando:
    - Su determinante exstá lejos de cero ($|\det(A)|$ significativo)
    - Su número de condición $\kappa(A)$ es relativamente bajo.

Estas matrices son estables para cálculos numéricos.


**Número de condición ($\kappa(A)$)**

El número de condición de una matriz se define como:
$$
\kappa(A) = \|A\| \cdot \|A^{-1}\|
$$
donde $\| \cdot \|$ es una norma matricial (típicamente la norma-2). Su interpretación es:

- $\kappa(A) \approx 1$: Matriz bien condicionada
- $\kappa(A) \gg 1$: Matriz mal condicionada
- $\kappa(A) \to \infty$: Matriz singular (no invertible)


El número de condición cuantifica la amplificación del error relativo al resolver sistemas lineales.

````python
# Definición de las matrices
A = np.array([[1, 2, 3],
              [2, 3, 4],
              [3, 4, 5]])

B = np.array([[2.11, -0.80, 1.72],
              [-1.84, 3.03, 1.29],
              [-1.57, 5.25, 4.30]])

C = np.array([[2, -1, 0],
              [-1, 2, -1],
              [0, -1, 2]])

D = np.array([[4, 3, -1],
              [7, -2, 3],
              [5, -18, 13]])
````

````python
# Función para clasificar una matriz
def clasificar_matriz(matriz, nombre):
    det = np.linalg.det(matriz)
    cond = np.linalg.cond(matriz)
    
    print(f"\nMatriz {nombre}:")
    print(f"Determinante = {det:.4f}")
    print(f"Número de condición (cond) = {cond:.4f}")
    
    if np.isclose(det, 0):
        print("Clasificación: SINGULAR (determinante = 0)")
    elif cond > 1e10:
        print("Clasificación: MAL CONDICIONADA")
    elif cond > 1e3:
        print("Clasificación: MODERADAMENTE MAL CONDICIONADA")
    else:
        print("Clasificación: BIEN CONDICIONADA")

# Clasificación de las matrices
clasificar_matriz(A, "A")
clasificar_matriz(B, "B")
clasificar_matriz(C, "C")
clasificar_matriz(D, "D")
````

# 4.

Escriba una función `my_is_orthogonal(v1, v2, tol)` donde:
- `v1` y `v2` son vectores columna del mismo tamaño
- `tol` es un valor escalar estrictamente mayor que cero.

La función debe devolver 1 si el ángulo $\theta$ entre `v1` y `v2` cumple que $\left|\frac{\pi}{2} - \theta\right| < \texttt{tol}$ y 0 en caso contrario.


````python
def my_is_orthogonal(v1, v2, tol):
    """
    Verifica si los vectores v1 y v2 son ortogonales dentro de una tolerancia dada.
    
    Parámetros:
    - v1, v2: Vectores columna (arrays de NumPy) del mismo tamaño.
    - tol: Tolerancia para la desviación del ángulo respecto a π/2 (debe ser > 0).
    
    Devuelve:
    - 1 si |π/2 - θ| < tol, donde θ es el ángulo entre v1 y v2.
    - 0 en caso contrario.
    """
    # Verificar que los vectores tengan el mismo tamaño
    if v1.shape != v2.shape:
        raise ValueError("Los vectores deben tener el mismo tamaño.")
    
    # Calcular el producto punto
    dot_product = np.dot(v1.flatten(), v2.flatten())
    
    # Calcular las normas de los vectores
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    
    # Evitar división por cero (si alguno de los vectores es cero, no se considera ortogonal)
    if norm_v1 == 0 or norm_v2 == 0:
        return 0
    
    # Calcular el coseno del ángulo θ
    cos_theta = dot_product / (norm_v1 * norm_v2)
    
    # Calcular θ (en radianes)
    theta = np.arccos(np.clip(cos_theta, -1.0, 1.0))  # Usamos clip para evitar errores numéricos
    
    # Verificar si |π/2 - θ| < tol
    if np.abs(np.pi/2 - theta) < tol:
        return 1
    else:
        return 0
````

````python
# Vectores casi ortogonales
v1 = np.array([1, 0.001])
v2 = np.array([0.001, 1])
tol = 0.01  # Tolerancia de 0.01 radianes (~0.57 grados)

print(my_is_orthogonal(v1, v2, tol))  # Salida: 1 (dentro de la tolerancia)
````

# 5.
Para dos matrices no singulares $A$ y $B$ verifique numéricamente la propiedad:

$$
det (A \cdot B)= det(A) \cdot det (B)
$$

````python
# Generar matrices aleatorias no singulares (o definir matrices específicas)
n = 3  # Tamaño de las matrices (3x3 como ejemplo)
A = np.random.rand(n, n)
B = np.random.rand(n, n)

# Asegurarse de que sean no singulares (det ≠ 0)
while np.linalg.det(A) == 0:
    A = np.random.rand(n, n)
while np.linalg.det(B) == 0:
    B = np.random.rand(n, n)

# Calcular det(A · B)
det_AB = np.linalg.det(A @ B)  

# Calcular det(A) * det(B)
det_A_times_det_B = np.linalg.det(A) * np.linalg.det(B)

# Tolerancia para comparación numérica
tolerance = 1e-10

# Verificar la propiedad
if np.abs(det_AB - det_A_times_det_B) < tolerance:
    print("La propiedad se cumple numéricamente:")
    print(f"det(A · B) = {det_AB:.6f}")
    print(f"det(A) * det(B) = {det_A_times_det_B:.6f}")
else:
    print("La propiedad no se cumple:")
    print(f"det(A · B) = {det_AB:.6f}")
    print(f"det(A) * det(B) = {det_A_times_det_B:.6f}")
````

Considere el sistema de tres masas y cuatro resortes mostrado en la figura.

Las ecuaciones de movimiento obtenidas a partir de $\sum F_x = ma_x$ para cada masa son:

$$
\ddot{x}_1 + \left(\frac{k_1}{m_1}\right)x_1 - \left(\frac{k_2}{m_1}\right)x_2 = 0 
$$
$$
\ddot{x}_2 - \left(\frac{k_2}{m_2}\right)x_1 + \left(\frac{k_2 + k_3}{m_2}\right)x_2 - \left(\frac{k_3}{m_2}\right)x_3 = 0 
$$
$$
\ddot{x}_3 - \left(\frac{k_3}{m_3}\right)x_2 + \left(\frac{k_3 + k_4}{m_3}\right)x_3 = 0
$$

donde:
- $x_i$ es el desplazamiento a la derecha de la masa $i$ desde su posición de equilibrio
- $k_1 = k_4 = 10 $ N/m
- $k_2 = k_3 = 30$ N/m
- $m_1 = m_2 = m_3 = 1$ Kg


El sistema puede escribirse en forma matricial como:
$$
[\text{vector aceleración } \mathbf{\ddot{x}}] + [k/m \text{ matriz}][\text{vector desplazamiento } \mathbf{x}]=0
$$

Para un instante particular donde $x_1 = 0.05 m, x_2 = 0.04m, x_3 = 0.03 m$, calcule el vector de aceleraciones $\begin{bmatrix} \ddot{x}_1 & \ddot{x}_2 & \ddot{x}_3 \end{bmatrix}^T$.



````python
# Constantes del problema (en unidades SI)
k1 = 10
k2 = 30
k3 = 30
k4 = 10
m1 = 1
m2 = 1
m3 = 1

# Matriz de coeficientes del sistema (k/m matrix)
K = np.array([
    [k1/m1, -k2/m1, 0],
    [-k2/m2, (k2 + k3)/m2, -k3/m2],
    [0, -k3/m3, (k3 + k4)/m3]
])

# Vector de desplazamientos en el instante dado (en metros)
x = np.array([0.05, 0.04, 0.03])

# Calculamos el vector de aceleraciones: ̈x = -K * x
x_ddot = -np.dot(K, x)

# Mostramos el resultado con 4 decimales
print("Vector de aceleraciones (m/s²):")
print(np.round(x_ddot, 4))
````
